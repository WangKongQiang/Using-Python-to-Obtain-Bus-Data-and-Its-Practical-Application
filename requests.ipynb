(1)requests是什么?

不建议使用Python自带的标准库 urllib urllib3

Requests 用于访问网络资源，是用 Python 语言编写，基于 urllib, 采用 Apache2 Licensed 开源协议的 HTTP 库，它比 urllib 更加方便，可以节约我们大量的工作，完全满足HTTP测试需求。
