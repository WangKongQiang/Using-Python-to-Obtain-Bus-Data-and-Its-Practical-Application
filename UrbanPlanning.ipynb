```python
from lxml import etree
import requests
import pandas as pd
def get_dt():
    dt=[]
    for i in range(1,2):
        # url = 'http://www.planning.org.cn/news/newslist?cid=12&page={}'.format(str(i))
        url = 'http://www.planning.org.cn/news/list?cid=12&page={}'.format(str(i))
        html = requests.get(url).text
        txt = etree.HTML(html)
        # txt2 = txt.xpath('//div[@class="fl w680 overh xh_list_boxb content"]/div/div')
        txt2 = txt.xpath('//div[@class="w900 overh content"]/div/div/div')
        # for t in txt2[1:]:
        for t in txt2[0:]:
            row=[]  # // 列表里有两个元素，一个是title, 一个是url地址。
            title =t.xpath('h4/a/text()')[0]
            url = t.xpath('h4/a/@href')[0]
            row.append(title)
            row.append('http://www.planning.org.cn'+url)
            dt.append(row)
    return dt
```

```python
pd.DataFrame(get_dt(), columns=['title','url'])
```

```python
from lxml import etree
import requests
import pandas as pd

def get_dt():
    dt=[]
    for i in range(1, 11):
        # url = 'http://www.planning.org.cn/news/newslist?cid=12&page={}'.format(str(i))
        url = 'http://www.planning.org.cn/news/list?cid=12&page={}'.format(str(i))
        html = requests.get(url).text
        txt = etree.HTML(html)
        # txt2 = txt.xpath('//div[@class="fl w680 overh xh_list_boxb content"]/div/div')
        txt2 = txt.xpath('//div[@class="w900 overh content"]/div/div/div')
        # for t in txt2[1:]:
        for t in txt2[0:]:
            row=[]
            title =t.xpath('h4/a/text()')[0]
            url = t.xpath('h4/a/@href')[0]
            row.append(title)
            row.append('http://www.planning.org.cn'+url)
            dt .append(row)
    return dt
```

```python
pd.DataFrame(get_dt(), columns=['title','url'])
```
